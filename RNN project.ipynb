```python
import kagglehub
path = kagglehub.dataset_download("jp797498e/twitter-entity-sentiment-analysis")
```

```python
import os
import pandas as pd

# List files in the directory to find the CSV file
file = os.listdir(path)[1]
path=os.path.join(path,file)
df=pd.read_csv(path)
df.head()
```

```python
df.info()
```

```python
df["Positive"].value_counts()
```

```python
df.drop(columns=["2401","Borderlands"],inplace=True)
```

```python
df.info()
```

```python
df.head()
```

```python
df.columns=["Sentiment","Tweets"]
```

```python
df.head()
```

```python
data=df.dropna()
```

```python
data.isnull().sum()
```

```python
data.drop_duplicates(inplace=True)
```

```python
data.duplicated().sum()
```

```python
data=data[data["Sentiment"]!="Irrelevant"]
```

```python
data["Sentiment"].value_counts()
```

```python
## Feature encoding
```

```python
data=data.sample(24000,random_state=42)
```

```python
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
data["Sentiment"]=le.fit_transform(data["Sentiment"])
```

```python
data.head()
```

```python
from tensorflow.keras.preprocessing.text import Tokenizer
```

```python
tokenizer=Tokenizer(oov_token="Nothing")
tokenizer.fit_on_texts(data["Tweets"])
```

```python
tokenizer.word_index
```

```python
tokenizer.word_counts
```

```python
tokenizer.document_count
```

```python
len(tokenizer.word_index)  ## Vocabulary size
```

```python
sequences=tokenizer.texts_to_sequences(data["Tweets"])
```

```python
sequences
```

```python
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

```python
sequences=pad_sequences(sequences=sequences,padding="pre")
```

```python
sequences
```

```python
from tensorflow.keras.layers import SimpleRNN,Dense,Input,Embedding
```

```python
from tensorflow.keras.models import Sequential
```

```python
X=sequences
Y=data["Sentiment"].values
```

```python
from sklearn.model_selection import train_test_split

X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.2,random_state=42)
```

## **Without Embedding Layer**

```python
model=Sequential()
model.add(SimpleRNN(32,input_shape=(99,1),return_sequences=False))
model.add(Dense(3,activation="softmax"))
```

```python
model.summary()
```

```python
model.compile(optimizer="adam",loss="sparse_categorical_crossentropy",metrics=["accuracy"])
```

```python
model.fit(X_train,Y_train,epochs=10,validation_data=(X_test,Y_test),batch_size=64)
```

## **With Embedding Layer**

```python
vocab_size=len(tokenizer.word_index)+1
vocab_size
```

```python
model_2=Sequential()
model_2.add(Input(shape=(99,)))
model_2.add(Embedding(input_dim=vocab_size,output_dim=128))
model_2.add(Dense(3,activation="softmax"))
```

```python
model_2.summary()
```

```python
model_final = Sequential([
    Input(shape=(117,)),
    Embedding(input_dim=vocab_size, output_dim=128),
    SimpleRNN(32, dropout=0.2, recurrent_dropout=0.2),
    Dense(3, activation="softmax")
])
```

```python
model_final.compile(optimizer="adam",loss="sparse_categorical_crossentropy",metrics=["accuracy"])
```

```python
model_final.summary()
```

```python
model_final.fit(X_train,Y_train,epochs=50,validation_data=(X_test,Y_test),batch_size=64)
```

```python
import pickle
```

```python
with open("tokenizer.pkl","wb")as file:
  pickle.dump(tokenizer,file)

model_final.save("rnn_model.h5")
```
